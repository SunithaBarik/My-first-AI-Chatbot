```python
# My First AI Chatbot
def chatbot():
    print("AI Bot: Hi! Mai tumhara AI Bot hu. 'bye' likhne pe band ho jaunga.")
    
    while True:
        user = input("Tum: ").lower()
        
        if "hello" in user or "hi" in user:
            print("AI Bot: Hello! Kaise ho?")
        elif "name" in user:
            print("AI Bot: Mera naam Mini-AI hai, tumne banaya hai!")
        elif "python" in user:
            print("AI Bot: Python AI ke liye best language hai.")
        elif "bye" in user:
            print("AI Bot: Bye Bye! Milte hai fir.")
            break
        else:
            print("AI Bot: Hmm, ye mujhe samajh nahi aaya, kuch aur pucho.")

chatbot()
```
